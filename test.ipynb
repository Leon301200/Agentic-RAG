```python
from sentence_transformers import SentenceTransformer
textes = [
    "The weather is lovely today.",
    "It's so sunny outside!",
    "He drove to the stadium.",
    ]
model = SentenceTransformer("intfloat/multilingual-e5-base")   # 768 dim, max 512 tokens
emb = model.encode(textes, normalize_embeddings=True)          # -> np.ndarray de forme [n, 768]
```

```python
emb.shape
```

```python
similarity = model.similarity(emb, emb)
print(similarity)
```
